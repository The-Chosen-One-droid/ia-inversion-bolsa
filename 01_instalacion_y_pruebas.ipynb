```python
# 1. Instalamos yfinance (pandas, numpy y matplotlib ya vienen preinstalados en Colab)
!pip install yfinance

# 2. Importamos las librerías clave
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf

print("¡Todas las librerías se han instalado e importado correctamente!")
```

```python
# Definimos el ticker de la acción (AAPL = Apple) y el rango de fechas
ticker = "AAPL"
fecha_inicio = "2020-01-01"
fecha_fin = "2026-01-01"

# Descargamos los datos financieros
df = yf.download(ticker, start=fecha_inicio, end=fecha_fin)

# Mostramos las primeras 5 filas de la tabla
print("--- Primeras 5 filas ---")
display(df.head())

# Mostramos la información básica del conjunto de datos
print("\n--- Información técnica del DataFrame ---")
df.info()
```
